Cell 1 - Load processed data

```python
from pathlib import Path
import pandas as pd
import numpy as np

BASE_DIR = Path(r"C:\Users\NickK\OneDrive\Υπολογιστής\github\kaggle_store_sales_forecasting")
PROCESSED_DIR = BASE_DIR / "data" / "processed"

train = pd.read_csv(PROCESSED_DIR / "train_features.csv")

train["date"] = pd.to_datetime(train["date"])

print(train.shape)
```

Cell 2 - Sort correctly

```python
train = train.sort_values(
    ["store_nbr", "family", "date"]
)

train.head()
```

Cell 3 - Create Lag 7

```python
train["sales_lag_7"] = (
    train.groupby(
        ["store_nbr", "family"]
    )["sales"]
    .shift(7)
)
```

Cell 4 - Create Lag 14

```python
train["sales_lag_14"] = (
    train.groupby(
        ["store_nbr", "family"]
    )["sales"]
    .shift(14)
)
```

Cell 5 - Create Lag 28

```python
train["sales_lag_28"] = (
    train.groupby(
        ["store_nbr", "family"]
    )["sales"]
    .shift(28)
)
```

Cell 6 - Check

```python
train[
    [
        "date",
        "store_nbr",
        "family",
        "sales",
        "sales_lag_7",
        "sales_lag_14",
        "sales_lag_28"
    ]
].head(40)
```

Cell 7 - Missing values

```python
print(train["sales_lag_7"].isnull().sum())
print(train["sales_lag_14"].isnull().sum())
print(train["sales_lag_28"].isnull().sum())
```

Cell 8 - Rolling Mean 7

```python
train["rolling_mean_7"] = (
    train.groupby(
        ["store_nbr", "family"]
    )["sales"]
    .transform(
        lambda x: x.shift(1).rolling(7).mean()
    )
)
```

Cell 9 - Rolling Mean 14

```python
train["rolling_mean_14"] = (
    train.groupby(
        ["store_nbr", "family"]
    )["sales"]
    .transform(
        lambda x: x.shift(1).rolling(14).mean()
    )
)
```

Cell 10 - Rolling Mean 28

```python
train["rolling_mean_28"] = (
    train.groupby(
        ["store_nbr", "family"]
    )["sales"]
    .transform(
        lambda x: x.shift(1).rolling(28).mean()
    )
)
```

Cell 11 - Check results

```python
train[
    [
        "date",
        "sales",
        "sales_lag_7",
        "rolling_mean_7",
        "rolling_mean_14",
        "rolling_mean_28"
    ]
].head(40)
```

Cell 12 - Missing values

```python
print(train["rolling_mean_7"].isnull().sum())
print(train["rolling_mean_14"].isnull().sum())
print(train["rolling_mean_28"].isnull().sum())
```

Cell 13 - Rolling Standard Deviation

```python
train["rolling_std_7"] = (
    train.groupby(
        ["store_nbr", "family"]
    )["sales"]
    .transform(
        lambda x: x.shift(1).rolling(7).std()
    )
)
```

Cell 14

```python
train["rolling_std_28"] = (
    train.groupby(
        ["store_nbr", "family"]
    )["sales"]
    .transform(
        lambda x: x.shift(1).rolling(28).std()
    )
)
```

Cell 15

```python
print(train["rolling_std_7"].isnull().sum())
print(train["rolling_std_28"].isnull().sum())
```

Cell 16 - Fill NaNs

```python
lag_features = [
    "sales_lag_7",
    "sales_lag_14",
    "sales_lag_28",
    "rolling_mean_7",
    "rolling_mean_14",
    "rolling_mean_28",
    "rolling_std_7",
    "rolling_std_28"
]

train[lag_features] = train[lag_features].fillna(0)
```

Cell 17 - Everything should be zero

```python
train[lag_features].isnull().sum()
```

Cell 18 - Save advanced dataset

```python
ADVANCED_DIR = BASE_DIR / "data" / "advanced"

ADVANCED_DIR.mkdir(
    parents=True,
    exist_ok=True
)

train.to_csv(
    ADVANCED_DIR / "train_advanced.csv",
    index=False
)

print(train.shape)
```
