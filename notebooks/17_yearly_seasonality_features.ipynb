Cell 1

```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path(
    r"C:\Users\NickK\OneDrive\Υπολογιστής\github\kaggle_store_sales_forecasting"
)

ADVANCED_DIR = BASE_DIR / "data" / "advanced"
```

Cell 2

```python
train = pd.read_csv(
    ADVANCED_DIR /
    "train_expanded_ewma.csv"
)

train["date"] = pd.to_datetime(
    train["date"]
)

print(train.shape)
```

Cell 3

```python
train = train.sort_values(
    ["store_nbr", "family", "date"]
).reset_index(drop=True)
```

Cell 4

```python
train["sales_lag_364"] = (
    train.groupby(
        ["store_nbr", "family"]
    )["sales"]
    .shift(364)
)

print("lag_364 done")
```

Cell 5

```python
train["sales_lag_365"] = (
    train.groupby(
        ["store_nbr", "family"]
    )["sales"]
    .shift(365)
)

print("lag_365 done")
```

Cell 6

```python
train["sales_lag_371"] = (
    train.groupby(
        ["store_nbr", "family"]
    )["sales"]
    .shift(371)
)

print("lag_371 done")
```

Cell 7

```python
train["rolling_mean_364_7"] = (
    train.groupby(
        ["store_nbr", "family"]
    )["sales"]
    .transform(
        lambda x:
        x.shift(364)
         .rolling(7)
         .mean()
    )
)

print("rolling_mean_364_7 done")
```

Cell 8

```python
seasonal_features = [
    "sales_lag_364",
    "sales_lag_365",
    "sales_lag_371",
    "rolling_mean_364_7"
]

train[seasonal_features] = (
    train[seasonal_features]
    .fillna(0)
)

print(
    train[seasonal_features]
    .isnull()
    .sum()
)
```

Cell 9

```python
train.to_csv(
    ADVANCED_DIR /
    "train_expanded_ewma_seasonal.csv",
    index=False
)

print(train.shape)
```
