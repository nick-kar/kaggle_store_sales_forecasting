Cell 1

```python
from pathlib import Path
import pandas as pd
import numpy as np

BASE_DIR = Path(
    r"C:\Users\NickK\OneDrive\Υπολογιστής\github\kaggle_store_sales_forecasting"
)

ADVANCED_DIR = BASE_DIR / "data" / "advanced"
```

Cell 2

```python
train = pd.read_csv(
    ADVANCED_DIR / "train_expanded_features.csv"
)

train["date"] = pd.to_datetime(
    train["date"]
)

print(train.shape)
```

Cell 3

```python
train = train.sort_values(
    ["store_nbr", "family", "date"]
).reset_index(drop=True)
```

Cell 4

```python
train["ewma_03"] = (
    train.groupby(
        ["store_nbr", "family"]
    )["sales"]
    .transform(
        lambda x:
        x.shift(1)
         .ewm(alpha=0.3)
         .mean()
    )
)

print("ewma_03 done")
```

Cell 5

```python
train["ewma_05"] = (
    train.groupby(
        ["store_nbr", "family"]
    )["sales"]
    .transform(
        lambda x:
        x.shift(1)
         .ewm(alpha=0.5)
         .mean()
    )
)

print("ewma_05 done")
```

Cell 6

```python
train["ewma_07"] = (
    train.groupby(
        ["store_nbr", "family"]
    )["sales"]
    .transform(
        lambda x:
        x.shift(1)
         .ewm(alpha=0.7)
         .mean()
    )
)

print("ewma_07 done")
```

Cell 7

```python
train[
    [
        "ewma_03",
        "ewma_05",
        "ewma_07"
    ]
] = train[
    [
        "ewma_03",
        "ewma_05",
        "ewma_07"
    ]
].fillna(0)

print(
    train[
        [
            "ewma_03",
            "ewma_05",
            "ewma_07"
        ]
    ].isnull().sum()
)
```

Cell 8

```python
train.to_csv(
    ADVANCED_DIR /
    "train_expanded_ewma.csv",
    index=False
)

print(train.shape)
```
